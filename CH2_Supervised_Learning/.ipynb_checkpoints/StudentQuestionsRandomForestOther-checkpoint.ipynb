## 1) 
Could you explain the difference between "random sampling with replacement" and "random sampling without replacement"?  "Random sampling" seems obvious, but what are we replacing?

A bootstrap sample is a random sample with replacement. So, it has the same size as the original sample but might duplicate some of the original observations. A sample in the case of bagged trees is a row of data. With bootstrapped your data may have duplicate rows. 

Bagging increases predictive accuracy by **reducing the variance**, similar to how cross-validation reduces the variance associated with train/test split (for estimating out-of-sample error) by splitting many times an averaging the results.

#### Sample with Replacement
Think of it as taking a marble out of the bag and putting it back in. Shuffling the bag, and taking out another one. It is entirely possible you could take out the same marble that you have already taken out. 

```python
for _ in range(1, 6):
    samples = np.random.choice(a=15, size=15, replace=True)
    print(samples, ',Num Samples: ' + str(len(samples)) + ',numberUnique: ' + str(len(set(samples))) )
```

#### Sample without Replacement
Think of it as taking a marble out of the bag and NOT putting it back in. Shuffling the bag, and taking out another one. It is not possible you could take out the same marble as before since it was already out of the bag. 

```python
for _ in range(1, 15):
    samples = np.random.choice(a=15, size=15, replace=False)
    print(samples, ',Num Samples: ' + str(len(samples)) + ',numberUnique: ' + str(len(set(samples))) )
```

## 2)
Could you define/explain "out-of-sample error"?

One of the goals of machine learning is to make a model that will generalize well on new data (also known as out of sample data). When we covered train-test-split, we looked at different evaluation procedures for predicting model out-of-sample accuracy/score.Another name for out-of-sample error is generalization error.

Error metrics should be computed on new data because in sample validation (predicting on your training data makes it a lot more likely you will overfit). As far as the word error, think of it in terms of regression problems. Often we have had MAE, MSE, or RMSE. For classification, often people use the term error for their metrics. 

One cool thing about bagged models, is that out-of-sample error can be estimated without using train/test split or cross-validation!

On average, each bagged tree uses about two-thirds of the observations. For each tree, the remaining observations are called "out-of-bag" observations.

```python
# Sample with replacement code
for _ in range(1, 6):
    samples = np.random.choice(a=15, size=15, replace=True)
    print(samples, ',Num Samples: ' + str(len(samples)) + ',numberUnique: ' + str(len(set(samples))) )
```

```python
# Sample with replacement code
for i in range(1, 6):
    samples = np.random.choice(a=15, size=15, replace=True)
    print('iteration: ' + str(i),samples, ',Num Samples: ' + str(len(samples)) + ',numberUnique: ' + str(len(set(samples))) )
    print('(Row) Numbers not in the sample: ', set(range(0,15)).difference(set(samples)))
    print('\n')
```

Side Note: 
    
How to calculate **"out-of-bag error":**

1. For every observation in the training data, predict its response value using **only** the trees in which that observation was out-of-bag. Average those predictions (for regression) or take a vote (for classification).
2. Compare all predictions to the actual response values in order to compute the out-of-bag error.

When N is sufficiently large, the **out-of-bag error** is an accurate estimate of **out-of-sample error**.    

3. The description of Random Forests says ".. each time a split is considered, a
random sample of m features is chosen as split candidates from the full set of p
features. The split is only allowed to use one of those m features. " So we choose a subset of features, then choose 1 feature from that subset. How is that different from just choosing 1 feature? Is it that the choice of subset is randomized, while the choice of 1 feature from that subset is based on something else?

The difference is that when building each tree, each time a split is considered, a random sample of m features is chosen as split candidates from the full set of p features. The split is only allowed to use one of those m features. 
A new random sample of features is chosen for every single tree at every single split. 

For classification, m is typically chosen to be the square root of p

For regression, m is typically chosen to be somewhere between p/3 and p. 

<b>Why do we do this? </b>

Suppose there is <b>one very strong feature</b> in the data set. When using bagged trees, most of the trees will use that feature as the top split, resulting in an ensemble of similar trees that are <b>highly correlated</b>. This is not a good thing. Averaging highly correlated quantities does not significantly reduce variance (which is the goal of bagged trees). By randomly leaving out candidate features from each split, <b>random forests “decorrelate” the trees</b> to the extent that the averaging process can reduce the variance of the resulting model. Having all the same trees means the voting will all predict the same thing.
