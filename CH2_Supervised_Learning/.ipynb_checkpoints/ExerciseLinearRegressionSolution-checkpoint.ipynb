## Linear Regression

In the file linear.csv there is a set of 100 x,y pairs of numbers in csv format.

About 10% of the lines have a missing y value.

Your goal is to do the following:

1. Preprocess the data to remove any points with a missing y value
2. Fit a linear regression model using scikit-learn's LinearRegression package
3. Create a scatter plot of the populated x,y pairs as blue points and the best fit linear regression line in red

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = pd.read_csv("data/linear.csv")
data.head()
```

```python
data['y'].isnull().sum()
```

#### 1. Preprocess the data to remove any points with a missing y value

```python
data
```

```python
data.dropna()
```

```python
# If you dont understand what I am doing, review the pandas lecture.
data = data.dropna(how='any')
```

#### 2. Fit a linear regression model using sklearn's LinearRegression

Subtask is to arrange data into a features matrix and target vector

```python
X = data['x'].values

# This is not the desired shape
X.shape
```

```python
X = X.reshape(92, 1)
```

```python
X.shape
```

```python
y = data['y'].values

# This could work
y.shape
```

```python
# I will reshape it anyways
y = y.reshape(92, 1)
```

```python
y.shape
```

```python
# Make an instance of LinearRegression

# Make a linear regression instance
lr = LinearRegression()

# Training the model on the data, storing the information learned from the data
# Model is learning the relationship between X and y 
lr.fit(X,y)

# Get Predictions for original x values
# This is not how we will do it for the rest of the course.
predictions = lr.predict(X)
```

#### 3. Plot a scatter plot of the populated x,y pairs as blue points and the best fit linear regression line in red

```python
plt.scatter(X, y, color='black');
plt.plot(X, predictions, color='red',linewidth=3);
```

#### What is the equation of the line for the regression? 

```python
lr.coef_
```

```python
lr.intercept_[0]
```

```python
m = lr.coef_[0][0]

b = lr.intercept_[0]

# following slope intercept form 
print("formula: y = {0}x + {1}".format(m, b) )
```
