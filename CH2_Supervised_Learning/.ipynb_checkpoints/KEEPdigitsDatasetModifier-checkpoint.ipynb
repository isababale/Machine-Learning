```python
%matplotlib inline

import matplotlib.pyplot as plt
import pandas as pd

from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
```

## Load the Dataset
 The code below loads a modified version of the iris dataset which has two classes. A 1 is a virginica flower and a 0 is not. In case you are curious, in this dataset a 0 is a versicolor flower. 

```python
data = load_digits()
```

```python
data = load_digits()
df = pd.DataFrame(data.data)
df['target'] = data.target
df.head()
```

```python
df.shape
```

## Transform the Dataset into a 4 Class Classiifcation Problem
Logistic regression is traditionally meant for binary classification 

```python
# Getting only the versicolor and virginica
removeSetosaFilter = df.loc[:, 'target'].isin([0,1,2, 3])
```

```python
df = df.loc[removeSetosaFilter,:]
```
